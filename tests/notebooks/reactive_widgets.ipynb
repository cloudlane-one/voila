# Reactive rendering test

```python
from ipywidgets import Text, VBox
```

## Static pretext

```python
print('cell 0')
```

## Single widget test

```python
print('cell 1 start')

text = Text(value="1", description='Text:', disabled=False) if "text" not in locals() else locals()["text"]
display(text)

print('cell 1 end')
```

```python
print("dynamic cell ", text.value)
```

## Stacked widget test

```python
print('cell 2 start')

text2 = Text(value="2", description='Text:', disabled=False) if "text2" not in locals() else locals()["text2"]
text3 = Text(value="3", description='Text:', disabled=False) if "text3" not in locals() else locals()["text3"]

display(VBox([text2, text3]))

print('cell 2 end')
```

```python
print("dynamic cell ", text2.value)
```

```python
print("dynamic cell ", text3.value)
```
